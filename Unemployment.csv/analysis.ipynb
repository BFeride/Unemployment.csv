```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

data = pd.read_csv("unemployment.csv")

data.columns = data.columns.str.strip()

print(data.head())
print(data.info())
print(data.isnull().sum())
data = data.dropna()

plt.figure(figsize=(8, 6))
sns.heatmap(data.select_dtypes(include=[np.number]).corr(), annot=True, cmap="coolwarm")
plt.title("Korrelyasiya Matrisi")
plt.show()

plt.figure(figsize=(10, 5))
sns.histplot(data["Estimated Unemployment Rate (%)"], kde=True, color="blue")
plt.title("İşsizlik Dərəcəsinin Paylanması")
plt.show()

data["Date"] = pd.to_datetime(data["Date"])
data["Year"] = data["Date"].dt.year

yearly_unemployment = (
    data.groupby("Year")["Estimated Unemployment Rate (%)"]
    .mean()
    .reset_index()
)

plt.figure(figsize=(10, 5))
sns.lineplot(
    data=yearly_unemployment,
    x="Year",
    y="Estimated Unemployment Rate (%)",
    marker="o",
    color="red",
)
plt.show()
```
